Q-3 Write a python function function to calculate the probability distribution function(PDF) of a Barnoulli distribution ?

```python
import scipy.stats as stats

def bernoulli_pmf(x, p):
    """
    Compute the probability mass function (PMF) of a Bernoulli distribution.

    Parameters:
    x : int (0 or 1) -> The possible outcome of the Bernoulli trial
    p : float (0 ≤ p ≤ 1) -> Probability of success (X = 1)

    Returns:
    float -> Probability of observing x
    """
    if x not in [0, 1]:  # Bernoulli outcomes must be 0 or 1
        raise ValueError("x must be either 0 or 1")
    if not (0 <= p <= 1):  # Probability must be between 0 and 1
        raise ValueError("p must be between 0 and 1")

    # PMF formula: P(X=x) = p^x * (1-p)^(1-x)
    return p**x * (1 - p)**(1 - x)

# Example Usage
p = 0.7  # Probability of success
x_values = [0, 1]

for x in x_values:
    print(f"P(X={x}) = {bernoulli_pmf(x, p):.4f}")
```

Q-9 Write a python function to calculate z-score from a dataset and plot them ?

```python
import numpy as np
import matplotlib.pyplot as plt
import scipy.stats as stats

def calculate_z_scores(data):
    """
    Calculate the Z-scores for a given dataset.

    Parameters:
        data (array-like): The dataset for which to compute Z-scores.

    Returns:
        z_scores (numpy array): The standardized Z-scores.
    """
    mean = np.mean(data)
    std_dev = np.std(data, ddof=1)

    z_scores = (data - mean) / std_dev
    return z_scores

np.random.seed(42)
sample_size = 1000
data = np.random.normal(loc=50, scale=10, size=sample_size)

z_scores = calculate_z_scores(data)

x = np.linspace(-4, 4, 1000)
pdf_values = stats.norm.pdf(x, 0, 1)

plt.figure(figsize=(8, 5))
plt.hist(z_scores, bins=30, density=True, alpha=0.6, color='blue', edgecolor='black', label="Z-score Histogram")

plt.plot(x, pdf_values, 'r-', lw=2, label="Standard Normal Curve")

plt.xlabel("Z-Score")
plt.ylabel("Density")
plt.title("Z-Score Distribution")
plt.axvline(x=0, color='black', linestyle='dashed', linewidth=1, label="Mean (Z=0)")
plt.legend()
plt.grid(True)

plt.show()
```

Q-10 Implement the central limit theoram (CLT) using python for a non-normal distribution ?

```python
import numpy as np
import matplotlib.pyplot as plt
import scipy.stats as stats

population_size = 100000
population = np.random.exponential(scale=2, size=population_size)

def sample_means(sample_size, num_samples):
    means = []
    for _ in range(num_samples):
        sample = np.random.choice(population, size=sample_size, replace=True)
        means.append(np.mean(sample))
    return np.array(means)

sample_sizes = [2, 10, 30, 50]
num_samples = 1000

plt.figure(figsize=(10, 8))

for i, sample_size in enumerate(sample_sizes, 1):
    sample_mean_distribution = sample_means(sample_size, num_samples)

    plt.subplot(2, 2, i)
    plt.hist(sample_mean_distribution, bins=30, density=True, alpha=0.6, color='blue', edgecolor='black', label="Sample Means")

    mu, sigma = np.mean(sample_mean_distribution), np.std(sample_mean_distribution)
    x = np.linspace(mu - 4*sigma, mu + 4*sigma, 1000)
    plt.plot(x, stats.norm.pdf(x, mu, sigma), 'r-', lw=2, label="Normal Approximation")

    plt.xlabel("Sample Mean")
    plt.ylabel("Density")
    plt.title(f"Sample Size = {sample_size}")
    plt.legend()
    plt.grid(True)

plt.suptitle("Central Limit Theorem Demonstration (Exponential Distribution)", fontsize=14)
plt.tight_layout(rect=[0, 0, 1, 0.96])
plt.show()
```

Q-19 Implement a hypothesis testing using z-statistics for a sample dataset  ?

```python
import numpy as np
import scipy.stats as stats

def z_test_hypothesis(sample, population_mean, population_std, alpha=0.05, alternative='two-tailed'):
    """
    Perform a hypothesis test using Z-statistics.

    Parameters:
    - sample: array-like, the sample data
    - population_mean: float, the known population mean (H0)
    - population_std: float, the known population standard deviation
    - alpha: float, significance level (default = 0.05)
    - alternative: str, type of test ('two-tailed', 'left-tailed', or 'right-tailed')

    Returns:
    - Z-score, p-value, and the decision (Reject or Fail to Reject H0)
    """
    sample_mean = np.mean(sample)
    sample_size = len(sample)

    z_score = (sample_mean - population_mean) / (population_std / np.sqrt(sample_size))

    if alternative == 'two-tailed':
        p_value = 2 * (1 - stats.norm.cdf(abs(z_score)))
    elif alternative == 'left-tailed':
        p_value = stats.norm.cdf(z_score)
    else:
        p_value = 1 - stats.norm.cdf(z_score)

    decision = "Reject H0" if p_value < alpha else "Fail to Reject H0"

    return z_score, p_value, decision

np.random.seed(42)
sample_data = np.random.normal(loc=50, scale=10, size=40)

population_mean = 52
population_std = 10
alpha = 0.05

z_score, p_value, decision = z_test_hypothesis(sample_data, population_mean, population_std, alpha, 'two-tailed')

print(f"Z-Score: {z_score:.2f}")
print(f"P-Value: {p_value:.5f}")
print(f"Decision: {decision}")
```

Q-20 Create confidence interval for a dataset using python and interpret the result ?

```python
import numpy as np
import scipy.stats as stats

def confidence_interval(data, confidence=0.95):
    """
    Calculate the confidence interval for a dataset.

    Parameters:
    - data: array-like, sample dataset
    - confidence: float, confidence level (default = 0.95 for 95% CI)

    Returns:
    - Lower and Upper bound of the confidence interval
    """
    sample_mean = np.mean(data)
    sample_std = np.std(data, ddof=1)
    n = len(data)

    se = sample_std / np.sqrt(n)

    z_critical = stats.norm.ppf((1 + confidence) / 2)

    margin_of_error = z_critical * se
    lower_bound = sample_mean - margin_of_error
    upper_bound = sample_mean + margin_of_error

    return lower_bound, upper_bound, sample_mean

np.random.seed(42)
data = np.random.normal(loc=50, scale=10, size=100)

ci_lower, ci_upper, sample_mean = confidence_interval(data)

print(f"Sample Mean: {sample_mean:.2f}")
print(f"95% Confidence Interval: ({ci_lower:.2f}, {ci_upper:.2f})")
```

Q-21 Generate data from a normal distribution,  then calculate and interpret the confidence interval for its mean. ?

```python
import numpy as np
import scipy.stats as stats
import matplotlib.pyplot as plt

def confidence_interval(data, confidence=0.95):
    """
    Calculate the confidence interval for a dataset.

    Parameters:
    - data: array-like, sample dataset
    - confidence: float, confidence level (default = 0.95 for 95% CI)

    Returns:
    - Lower and Upper bound of the confidence interval
    """
    sample_mean = np.mean(data)
    sample_std = np.std(data, ddof=1)
    n = len(data)

    se = sample_std / np.sqrt(n)

    z_critical = stats.norm.ppf((1 + confidence) / 2)

    margin_of_error = z_critical * se
    lower_bound = sample_mean - margin_of_error
    upper_bound = sample_mean + margin_of_error

    return lower_bound, upper_bound, sample_mean

np.random.seed(42)
data = np.random.normal(loc=50, scale=10, size=100)

ci_lower, ci_upper, sample_mean = confidence_interval(data)

print(f"Sample Mean: {sample_mean:.2f}")
print(f"95% Confidence Interval: ({ci_lower:.2f}, {ci_upper:.2f})")

plt.figure(figsize=(8,5))
plt.hist(data, bins=15, alpha=0.6, color='b', edgecolor='black', density=True)
plt.axvline(sample_mean, color='red', linestyle='dashed', linewidth=2, label=f"Mean = {sample_mean:.2f}")
plt.axvline(ci_lower, color='green', linestyle='dashed', linewidth=2, label=f"CI Lower = {ci_lower:.2f}")
plt.axvline(ci_upper, color='green', linestyle='dashed', linewidth=2, label=f"CI Upper = {ci_upper:.2f}")

plt.title("Confidence Interval for Normal Distribution Mean")
plt.xlabel("Value")
plt.ylabel("Density")
plt.legend()
plt.grid()
plt.show()
```

```python
import numpy as np
import scipy.stats as stats
import matplotlib.pyplot as plt

mu = 50
sigma = 10

x = np.linspace(mu - 4*sigma, mu + 4*sigma, 1000)

pdf_values = stats.norm.pdf(x, loc=mu, scale=sigma)

plt.figure(figsize=(8,5))
plt.plot(x, pdf_values, color='blue', linewidth=2, label=f'N({mu}, {sigma}²)')
plt.xlabel('X')
plt.ylabel('Probability Density')
plt.title('Probability Density Function (PDF) of a Normal Distribution')
plt.axvline(mu, color='red', linestyle='dashed', linewidth=2, label=f'Mean (μ) = {mu}')
plt.fill_between(x, pdf_values, alpha=0.2, color='blue')
plt.legend()
plt.grid()
plt.show()
```

Q-23 Use python to calculate and interpret the cumulative distribution functon (CDF) of a poisson distribution ?

```python
import numpy as np
import scipy.stats as stats
import matplotlib.pyplot as plt

lambda_ = 5

x = np.arange(0, 20)

cdf_values = stats.poisson.cdf(x, mu=lambda_)

plt.figure(figsize=(8,5))
plt.step(x, cdf_values, where='mid', color='blue', label=f'Poisson(λ={lambda_})')
plt.xlabel('Number of Events (k)')
plt.ylabel('Cumulative Probability')
plt.title('Cumulative Distribution Function (CDF) of a Poisson Distribution')
plt.axvline(lambda_, color='red', linestyle='dashed', linewidth=2, label=f'λ (Mean) = {lambda_}')
plt.legend()
plt.grid()
plt.show()
```

Q-26 Calculate the range and interquartile range(IQR) of a dataset generated from a normal distribution ?

```python
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

np.random.seed(42)
data = np.random.normal(loc=50, scale=15, size=1000)

data_range = np.max(data) - np.min(data)

Q1 = np.percentile(data, 25)
Q3 = np.percentile(data, 75)
IQR = Q3 - Q1

print(f"Range: {data_range:.2f}")
print(f"Interquartile Range (IQR): {IQR:.2f}")

plt.figure(figsize=(8,5))
sns.boxplot(data, color='skyblue')

plt.text(Q1, 0.02, f'Q1 = {Q1:.2f}', horizontalalignment='center', fontsize=12, color='blue')
plt.text(Q3, 0.02, f'Q3 = {Q3:.2f}', horizontalalignment='center', fontsize=12, color='red')

plt.title("Boxplot of Normally Distributed Data with IQR")
plt.xlabel("Value")
plt.grid()
plt.show()
```

Q-27 Implement z-score normalization on a dataset and visualize its transformation ?

```python
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

np.random.seed(42)
data = np.random.normal(loc=50, scale=15, size=1000)

mean = np.mean(data)
std_dev = np.std(data, ddof=1)
normalized_data = (data - mean) / std_dev

print(f"Original Data -> Mean: {mean:.2f}, Std Dev: {std_dev:.2f}")
print(f"Normalized Data -> Mean: {np.mean(normalized_data):.2f}, Std Dev: {np.std(normalized_data, ddof=1):.2f}")

plt.figure(figsize=(10,5))

plt.subplot(1,2,1)
sns.histplot(data, bins=30, kde=True, color='blue')
plt.axvline(mean, color='red', linestyle='dashed', linewidth=2, label=f'Mean: {mean:.2f}')
plt.xlabel("Original Values")
plt.ylabel("Frequency")
plt.title("Original Distribution")
plt.legend()

plt.subplot(1,2,2)
sns.histplot(normalized_data, bins=30, kde=True, color='green')
plt.axvline(0, color='red', linestyle='dashed', linewidth=2, label='Mean: 0 (Standardized)')
plt.xlabel("Z-Score Values")
plt.ylabel("Frequency")
plt.title("Z-Score Normalized Distribution")
plt.legend()

plt.tight_layout()
plt.show()
```

Q-28 Write a python function to calculate the skewness and kurtosis of a dataset generated from a normal distribution ?

```python
import numpy as np
import scipy.stats as stats
import matplotlib.pyplot as plt
import seaborn as sns

def compute_skewness_kurtosis(data):
    """Calculates and prints the skewness & kurtosis of a dataset."""
    skewness = stats.skew(data)
    kurtosis = stats.kurtosis(data, fisher=True)

    print(f"Skewness: {skewness:.4f}")
    print(f"Kurtosis (Excess): {kurtosis:.4f}")

    return skewness, kurtosis

np.random.seed(42)
data = np.random.normal(loc=50, scale=15, size=1000)

skewness, kurtosis = compute_skewness_kurtosis(data)

plt.figure(figsize=(8,5))
sns.histplot(data, bins=30, kde=True, color='blue')
plt.axvline(np.mean(data), color='red', linestyle='dashed', linewidth=2, label='Mean')
plt.xlabel("Value")
plt.ylabel("Frequency")
plt.title(f"Histogram of Normally Distributed Data\nSkewness: {skewness:.4f}, Kurtosis: {kurtosis:.4f}")
plt.legend()
plt.grid()
plt.show()
```
